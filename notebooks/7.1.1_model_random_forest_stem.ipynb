# 7.1.1 - Model: Random Forest (stemmed variant)

**Training and evaluation of the Random Forest classifier on the stemmed variant of the TF-IDF data.**

## Theoretical basis

Random Forest is an ensemble method - instead of training a single tree (as in `6.1`), it trains many independent Decision Tree models and combines their predictions into one final decision in order to obtain a more stable and accurate classifier.

Each tree in the forest is built in a slightly different way from a regular Decision Tree:
- Bootstrap sampling (bagging) - each tree is trained on its own random sample of the TRAIN set, drawn with replacement (the same example can appear multiple times, while some examples may be completely omitted)
- Random subset of features per split - at each node split, the tree does not consider all features (as a regular Decision Tree does), but only a randomly selected subset (controlled by the `max_features` parameter)
- Multiple trees – the number of trees in the forest is determined by the `n_estimators` parameter (a larger number of trees generally produces more stable results, but increases training time).

The final prediction of the entire forest is a majority vote of all individual trees.

Why this helps:
- An individual tree is unstable - a small change in the training data can produce a completely different tree, and it is prone to overfitting
- By averaging over many different, weakly correlated trees, the errors of individual trees cancel each other out, and the forest as a whole generalizes better to new data

Cost:
- We lose the direct interpretability of an individual tree (now we have many trees instead of one, so we cannot simply "follow" a single sequence of decisions)
- We retain some degree of interpretability through `feature_importances_` (averaged across all trees in the forest, instead of coming from a single tree as in `6.1`)

## Note about the name `max_features`

`max_features` here is not the same parameter as `max_features` from `3_vectorization.ipynb` (there it controlled the size of the entire TF-IDF vocabulary).

`max_features` here controls how many features are randomly considered at each individual node split within a single tree.

## Processing flow:

1. We load the previously prepared stemmed TF-IDF data using the `load_variant()` function

2. We create a `RandomForestClassifier` and define the values of the `n_estimators`, `max_depth`, `max_features`, and `min_samples_leaf` hyperparameters that `GridSearchCV` will test

3. We call the shared `train_and_evaluate()` function:
    1. GridSearchCV on the TRAIN set:
        - tests all combinations of hyperparameters from the param_grid
        - uses k-fold cross-validation (5 folds by default)
        - selects the hyperparameter combination that gives the best F1 score
    2. Final model:
        - after selecting the best hyperparameters, GridSearchCV trains the best model on the entire TRAIN set
    3. Evaluation:
        - the best model is used on the TEST set for the first time
        - accuracy, precision, recall, and F1 are calculated, and a confusion matrix is created
    4. Saving the results

4. We analyze `feature_importances_` (averaged across all trees in the forest) to see which terms contribute the most to the decisions


```python
import sys
sys.path.append("..")

import numpy as np
import joblib

from src.data_io import load_variant
from src.modeling import train_and_evaluate

from sklearn.ensemble import RandomForestClassifier
```

## Loading the data

We load the stemmed variant (TF-IDF matrices + labels), saved in `3_vectorization.ipynb`.


```python
X_train, X_test, y_train, y_test = load_variant("stemmed")

print(f"Train: {X_train.shape}")
print(f"Test:  {X_test.shape}")
```

## Model and hyperparameters

Hyperparameters: `n_estimators`, `max_depth`, `max_features` and `min_samples_leaf`

- `n_estimators` - number of trees in the forest
  - More trees → more stable, reliable results (averaging over more "votes")
  - But also slower training - each additional tree increases training time

- `max_depth` - maximum depth of each individual tree in the forest (`None` → no explicit depth limit)

- `max_features` - how many features are randomly considered at each split
  - `"sqrt"` - considers √66.682 ≈ 258 features per split
  - `"log2"` - considers log₂(66.682) ≈ 17 features per split
  - Fewer features per split means more diverse trees (a lower chance that two trees make the same split), which can improve the ensemble effect

- `min_samples_leaf` - minimum number of examples that a leaf must contain for a split to be allowed
  - A larger value prevents the creation of very small leaves and thus provides additional regularization to the model


```python
model = RandomForestClassifier(random_state=1)

param_grid = {
    "n_estimators": [1000, 1500, 2000, 2500, 3000],
    "max_depth": [300, 400, 500, 600],
    "max_features": ["sqrt", "log2"],
    "min_samples_leaf": [1, 2, 3],
}
```

## Training and evaluation

Calls `train_and_evaluate()` from `src/modeling.py`.

The function tests all specified combinations of `n_estimators`, `max_depth`, `max_features` and `min_samples_leaf` values using 5-fold GridSearchCV on the train set, selects the combination with the best F1 score, then trains the best model on the entire train set and evaluates it on the test set.

The results are saved as the model, metrics, and confusion matrix.

Duration:
- Decision Tree (`6.1`) took several minutes for dozens of hyperparameter combinations (grid × 5-fold).
- Random Forest builds a larger number of trees FOR each combination, instead of just one
- Although each individual tree is cheaper to build here (it considers only a subset of features per split, not all of them), the total time will be significantly longer.


```python
metrics = train_and_evaluate(
    model=model,
    param_grid=param_grid,
    X_train=X_train,
    y_train=y_train,
    X_test=X_test,
    y_test=y_test,
    model_name="random_forest",
    variant="stem",
)
```

## Most influential terms (feature importance)

`feature_importances_` represents a measure of feature importance.

The value shows how much a particular term contributed to reducing "impurity" across all splits (Same as in `6.1` - `feature_importances_` has no sign).

The difference compared to `6.1`: here, that value is averaged across all trees in the forest, rather than coming from a single tree.

We expect the ensemble to "discover" a larger number of useful terms than an individual tree (with a single tree, only a small portion of the vocabulary was used - with hundreds of trees and a random subset of features considered at each split, significantly more terms will likely receive non-zero importance).

A higher value means that the feature made a greater contribution to reducing impurity in the splits of the Random Forest model's trees.

Feature importance does not show the direction of influence, meaning it does not indicate whether a term contributes to positive or negative sentiment. Therefore, here we obtain a single ranking of the most important terms.

We load the saved model and vectorizer back from disk in order to associate the `feature_importances_` values with specific terms.


```python
best_model = joblib.load("../models/random_forest_stem.joblib")
vectorizer = joblib.load("../models/tfidf_vectorizer_stem.joblib")

feature_names = np.array(vectorizer.get_feature_names_out())
importances = best_model.feature_importances_

top_idx = np.argsort(importances)[-20:][::-1]

print("Top 20 most influential terms:")
for idx in top_idx:
    print(f"  {feature_names[idx]:30s}  {importances[idx]:.4f}")

n_zero_importance = np.sum(importances == 0)
print(f"\nNumber of terms that the forest never used for splitting: {n_zero_importance} out of {len(importances)} ({n_zero_importance / len(importances):.1%})")
```

## Conclusions

- Best `max_depth`: 500, `max_features`: log2, `min_samples_leaf`: 1 and `n_estimators`: 2500

- Results:
  - Accuracy  : 0.8792
  - Precision : 0.8546
  - Recall    : 0.9150
  - F1        : 0.8838

- `feature_importances_` shows that the Random Forest used the entire TF-IDF vocabulary

- Most influential terms:
  - Terms that are expected appear, such as: bad, worst, great, wast,...
  - Interestingly, terms appear that would not be expected to be among the most influential, such as: movi, act, would

- Interestingly, the log2 limitation on the number of features was selected, which increases the diversity of the trees and contributes to the ensemble effect

- Compared to a single Decision Tree, using a larger number of trees enables a more stable model and reduces the problem of overfitting, because the final decision is based on a combination of predictions from a large number of different trees
